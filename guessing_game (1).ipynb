```python
import random
def play_games():
    print("This is the guessing game!")
    print("\nChoose difficulty:")
    print("1. easy")  #1-50
    print("2. medium")  #1-100
    print("3. hard")  #1-500

    try:
        choice = int(input("Enter chosen difficulty: "))
    except ValueError:
        print("Invalid input. Defaulting to medium difficulty")
        choice = 2

    if choice == 1:
        number = random.randint(1,50)
    elif choice == 2:
        number = random.randint(1,100)
    else:
        number = random.randint(1,500)

    attempts = 0

    while True:
        try:
            guess = int(input("Guess the number: "))
        except ValueError:
            print("Please enter a valid integer")
            continue

        attempts += 1

        if guess < number:
            print("Too low")
        elif guess > number:
            print("Too high")
        else:
            print(f"Correct! You guessed it in {attempts} attempts.")
            break

play_games()
```
